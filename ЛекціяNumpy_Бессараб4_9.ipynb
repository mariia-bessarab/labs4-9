## 19. Підсумкові тренувальні завдання

1. Створіть матрицю `X` форми `(5, 4)` із випадкових чисел.
2. Обчисліть середнє та стандартне відхилення кожної ознаки.
3. Стандартизуйте матрицю через broadcasting.
4. Виберіть рядки, у яких значення першої ознаки додатне.
5. Додайте до матриці новий стовпець одиниць.
6. Створіть вектор ваг і обчисліть прогноз `X @ w`.
7. Для заданого `y_true` обчисліть MSE.
8. Перевірте, чи немає в масивах `NaN` або `inf`.


```python
import numpy as np
```

```python
# 1.
np.random.seed(42)

X = np.random.randn(5, 4)

print("Матриця X:")
print(X)
print("Форма X:", X.shape)
```

```python
# 2.
mean = X.mean(axis=0)
std = X.std(axis=0)

print("\nСереднє кожної ознаки:")
print(mean)

print("\nСтандартне відхилення кожної ознаки:")
print(std)
```

```python
# 3.
X_std = (X - mean) / std

print("\nСтандартизована матриця:")
print(X_std)
```

```python
# 4.
positive_rows = X[X[:, 0] > 0]

print("\nРядки, де перша ознака додатна:")
print(positive_rows)
```

```python
# 5.
X_with_ones = np.column_stack((X, np.ones(X.shape[0])))

print("\nМатриця з новим стовпцем одиниць:")
print(X_with_ones)
```

```python
# 6.
w = np.array([0.5, -0.2, 0.8, 0.3, 1.0])

y_pred = X_with_ones @ w

print("\nВектор ваг:")
print(w)

print("\nПрогноз X @ w:")
print(y_pred)
```

```python
# 7.
y_true = np.array([1.2, -0.5, 0.8, 1.5, -1.0])

mse = np.mean((y_true - y_pred) ** 2)

print("\nMSE:")
print(mse)
```

```python
# 8.
print("\nПеревірка X на NaN:")
print(np.isnan(X).any())

print("Перевірка X на inf:")
print(np.isinf(X).any())

print("Перевірка прогнозу на NaN:")
print(np.isnan(y_pred).any())

print("Перевірка прогнозу на inf:")
print(np.isinf(y_pred).any())
```
