### Завдання 25
Дано два вектори:

```text
a = [1, 2]
b = [2, 1]
```

Обчисліть косинус кута між ними за формулою через скалярний добуток і норми векторів.

Додатково знайдіть сам кут у градусах.

````python
a = np.array([1, 2])
b = np.array([2, 1])

dot_product = np.dot(a, b)
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

cos_angle = dot_product / (norm_a * norm_b)
angle_rad = np.arccos(cos_angle)
angle_deg = np.degrees(angle_rad)

print("Косинус кута:", cos_angle)
print("Кут у градусах:", angle_deg)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

mean = np.mean(A, axis=0)
std = np.std(A, axis=0)

normalized = (A - mean) / std

print("Середнє кожного стовпця:", mean)
print("Стандартне відхилення:", std)

print("Нормалізована матриця:")
print(normalized)
````

````python
np.random.seed(42)

grades = np.random.randint(0, 101, 30)

average_grade = np.mean(grades)
median_grade = np.median(grades)
max_grade = np.max(grades)
min_grade = np.min(grades)

count_60 = np.sum(grades >= 60)

percent_90 = np.mean(grades >= 90) * 100

print("Оцінки:", grades)
print("Середня оцінка:", average_grade)
print("Медіана:", median_grade)
print("Найвища оцінка:", max_grade)
print("Найнижча оцінка:", min_grade)
print("Кількість студентів з оцінкою >= 60:", count_60)
print("Відсоток студентів з оцінкою >= 90:", percent_90, "%")
````

### Завдання 36
Дано результати продажів трьох товарів за чотири місяці:

```text
sales = [[120, 150, 170, 160],
         [80,  90,  100, 110],
         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

````python
sales = np.array([
    [120, 150, 170, 160],
    [80, 90, 100, 110],
    [200, 180, 210, 230]
])

product_totals = np.sum(sales, axis=1)
monthly_totals = np.sum(sales, axis=0)

best_product = np.argmax(product_totals)
best_month = np.argmax(monthly_totals)

average_product = np.mean(sales, axis=1)

print("Загальні продажі кожного товару:", product_totals)
print("Загальні продажі за місяцями:", monthly_totals)
print("Товар з найбільшими продажами: товар", best_product + 1)
print("Місяць з найбільшими продажами:", best_month + 1)
print("Середні продажі кожного товару:", average_product)
````

### Завдання 37
Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
np.random.seed(42)

matrix37 = np.random.randint(1, 21, (5, 5))

diagonal_main = np.trace(matrix37)
diagonal_secondary = np.trace(np.fliplr(matrix37))

row_max = np.max(matrix37, axis=1)
column_min = np.min(matrix37, axis=0)

mean_matrix = np.mean(matrix37)

matrix37_modified = matrix37.copy()
matrix37_modified[matrix37_modified < mean_matrix] = 0

print("Початкова матриця:")
print(matrix37)

print("Сума головної діагоналі:", diagonal_main)
print("Сума побічної діагоналі:", diagonal_secondary)
print("Максимуми кожного рядка:", row_max)
print("Мінімуми кожного стовпця:", column_min)
print("Середнє значення:", mean_matrix)

print("Матриця після заміни:")
print(matrix37_modified)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
A = np.array([
    [1, 2, 1],
    [2, 1, -1],
    [3, -1, 2]
])

b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)

check = A @ x

print("Розв'язок системи:", x)
print("A @ x:", check)
print("Вектор b:", b)

print("Результати збігаються:",
      np.allclose(check, b))
````
